```python
import pandas as pd
```

```python
df = pd.read_excel(r"D:\Guvi\Project 2 Agriculture\ICRISAT-District Level Data.xlsx")
df
```

```python
print(df.shape)      
print(df.columns)    
print(df.info()) 
```

```python
df.describe()
```

```python
df.columns = (
    df.columns.str.strip()
              .str.upper()
              .str.replace(" ", "_")
              .str.replace("(", "", regex=False)
              .str.replace(")", "", regex=False)
)
df.columns
```

```python
df.isnull().sum()
```

```python
df["DIST_CODE"] = df["DIST_CODE"].astype("object")
df["STATE_CODE"] = df["STATE_CODE"].astype("object")
df.info()
```

```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
print((df == -1).sum())
```

```python
df = df.replace(-1, 0)
print((df == -1).sum())
```

```python
df.to_excel(r"clean_agri_data.xlsx", index=False)
```
